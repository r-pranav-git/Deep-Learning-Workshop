Scholarship Prediction using ANN

```python
import tensorflow as tf
import numpy as np
#Input
X=np.array([
    [85,95,20,80],
    [60,75,60,50],
    [90,98,15,90],
    [55,60,80,45],
    [78,88,30,70],
    [40,55,90,35]
],dtype=float)
#Output
y=np.array([
    1,
    0,
    1,
    0,
    1,
    0
])
#Normalization - between 0 and 1 -for better output
X=X/100
#Creating model
model=tf.keras.Sequential([
    #Creating hidden layer
    tf.keras.layers.Dense(
        3,
        activation="relu",
        input_shape=(4,)
    ),
    #Creating output layer
    tf.keras.layers.Dense(
        1,
        activation="sigmoid"
    )
])
#How to compile
model.compile(
    optimizer='adam',
    loss="binary_crossentropy",
    metrics=['accuracy']
)
#How to train model
model.fit(
    X,
    y,
    epochs=100,
    verbose=1
)
#Test data
student=np.array([[82,94,18,75]])/100
prediction=model.predict(student)
print(prediction)

if prediction>0.5:
  print("Scholarship Approved")
else:
  print("Scholarship Rejected")
model.summary
```
